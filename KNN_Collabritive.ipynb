```python
sample_data = df.sample(frac=0.001)
```

## 🟢 Making Pivot Table for our Dataset

```python
Movies_item_matrix = sample_data.pivot_table(values="rating",index="title",columns="userId",aggfunc="mean",fill_value=0)
Movies_item_matrix.head()
```

```python
users_item_matrix = sample_data.pivot_table(values="rating",columns="title",index="userId",aggfunc="mean",fill_value=0)
users_item_matrix.head()
```

## 🟢 Get Nearest Moveis by **`K-Nearest Neighbors`** Algorithm

```python
movies_model_knn = NearestNeighbors(metric = "cosine", algorithm = "brute")
movies_model_knn.fit(Movies_item_matrix.values)
```

```python
users_model_knn = NearestNeighbors(metric = "cosine", algorithm = "brute")
users_model_knn.fit(users_item_matrix.values)
```

```python
def Get_Recommandation(id:int,Pivot_Table:pd.DataFrame,model:NearestNeighbors):


    query_index = id
    distances, indices = model.kneighbors(Pivot_Table.iloc[query_index,:].values.reshape(1, -1), n_neighbors = 10)

    for i in range(0, len(distances.flatten())):
        if i == 0:
            print(f'Recommendations for {Pivot_Table.index[query_index]}:\n')
        else:
            print(f'{i}: {Pivot_Table.index[indices.flatten()[i]]}, with distance of {distances.flatten()[i]}:')
```

## 📍 Item-Based Collaborative Filtering
Your implementation uses a specific branch called **Item-Item Collaborative Filtering**.

* **The Logic**: Instead of looking for "Users who are like you," the model looks for "Movies that are liked by the same people who liked this movie".
* **The Benefit**: Item-item relationships are generally more stable than user preferences, making the recommendations more reliable over time.


```python
Get_Recommandation(id=1,Pivot_Table=Movies_item_matrix,model=movies_model_knn)
```

## 📍 User-Based Collaborative Filtering

* **The Logic**: Instead of looking for "Movies similar to Toy Story," the model looks for **"Users who have the same taste as you."** If User A and User B both gave 5 stars to the same 10 movies, the model assumes they are "Nearest Neighbors" and recommends movies User B liked that User A hasn't seen yet.
* **The Benefit**: It is excellent for **discovery**. It can recommend a movie from a completely different genre simply because someone with your exact "taste profile" enjoyed it, whereas Item-Based filtering often keeps you within the same category.

```python
Get_Recommandation(id=1,Pivot_Table=users_item_matrix,model=users_model_knn)
```
